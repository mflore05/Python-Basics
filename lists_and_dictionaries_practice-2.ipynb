## Problem 2: GC-Content Classifier
Calculate the GC-content percentage for a list of DNA reads and filter for high-GC sequences (>50%).
Input data:
    reads = ["ATGCATGC", "GCGCGCGC", "ATATATAT", "CGGCGGAT", "TTATATAA"]
Tasks:
* Write a function or comprehension that calculates the GC percentage: ((Count of G + Count of C)/ Total length) * 100
* Filter the list to include only reads where the GC content is strictly greater than 50.0%.
* Store the result in a dictionary named high_gc_reads where the key is the sequence and the value is its rounded GC percentage.

```python
reads = ["ATGCATGC", "GCGCGCGC", "ATATATAT", "CGGCGGAT", "TTATATAA"]

# Let's write and define a function that calculates the GC %
def calc_gc(seq):
    return ((seq.count("G") + seq.count("C")) / len(seq)) * 100

# Let's build the dictionary. Loop through each read one by one.
high_gc_reads = {}

for read in reads:
    if calc_gc(read) > 50.0:
        high_gc_reads[read] = round(calc_gc(read), 1)

print(high_gc_readss
```

## Problem 3: FASTQ Header Parser
Process raw sequencing header strings to extract metadata into a structured dictionary.
Input data:
    headers = [
    "SEQ01_len150_QCpass",
    "SEQ02_len75_QCfail",
    "SEQ03_len150_QCpass",
    "SEQ04_len100_QCpass",
    "SEQ05_len50_QCfail"
]
Tasks:
* Use comprehension to filter out only headers ending with QCpass.
* Split the string to isolate the sample name (e.g., 'SEQ01') and add the integer length (e.g., 150).
* Create a dictionary named passing_qc mapping each passing sample name to its length as an integer.

```python
headers = [
    "SEQ01_len150_QCpass",
    "SEQ02_len75_QCfail",
    "SEQ03_len150_QCpass",
    "SEQ04_len100_QCpass",
    "SEQ05_len50_QCfail"
]

passing_qc = {
    h.split("_")[0]: int(h.split("_")[1].replace("len", ""))
    for h in headers
    if h.endswith("QCpass")
}

print(passing_qc)
```

## Problem 5: Gene Expression Threshold Filter
Parse gene expression entries and retain only moderately-to-highly expressed genes (exp > 20.0).
Input data:
    genes = [
    "ENSG01_TP53_exp:45.2",
    "ENSG02_EGFR_exp:12.8",
    "ENSG03_MYC_exp:88.6",
    "ENSG04_BRCA1_exp:5.4",
    "ENSG05_GAPDH_exp:150.0"
]
Complete the following tasks:
1. Split each string by _ to isolate the gene symbol ('TP53') at index [1] and expression tag ('exp:45.2') at index [2].
2. Extract the expression value and convert it into a float (e.g., 45.2).
3. Filter to include only genes where the expression level is strictly greater than 20.0.
4. Create a dictionary named high_expression mapping the gene symbol to its float expression value.

```python
genes = [
    "ENSG01_TP53_exp:45.2",
    "ENSG02_EGFR_exp:12.8",
    "ENSG03_MYC_exp:88.6",
    "ENSG04_BRCA1_exp:5.4",
    "ENSG05_GAPDH_exp:150.0"
]

high_expression = {
    g.split("_")[1]: float(g.split("_")[2].replace("exp:", ""))
    for g in genes
    if float(g.split("_")[2].replace("exp:", "")) > 20.0
}

print(high_expression)
```
