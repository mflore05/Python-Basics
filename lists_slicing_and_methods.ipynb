Exercise 1: Given the list of recorded sensor readings:
    readings = [12.4, 15.1, 19.8, 22.0, 25.3, 28.9, 31.2]
Complete the following tasks:

```python
# Task 1: Slice the first 3 readings
readings = [12.4, 15.1, 19.8, 22.0, 25.3, 28.9, 31.2]
first_three = readings[0:3]
print("First Part:", first_three)

# Task 2: Slice the first 2 readings using negative indices
last_two = readings[-2:]
print("Last Two:", last_two)

# Task 3: Slice every second reading across the entire list
every_second = readings[::2]
print("Every Second:", every_second)
```

Exercise 2: Given this queue of sample IDs
    samples = ["sample_A", "sample_B", "corrupted", "sample_D"]
Solve the following tasks: 

```python
# Task 1: Remove corrupted from the list using .remove()
samples = ["sample_A", "sample_B", "corrupted", "sample_D"]
samples.remove("corrupted")

# Task 2: Add "sample_E" to the very end of the list using .append()
samples.append("sample_E")

# Task 3: Insert "sample_C" at index 2 so the list is in alphabetical order using .insert()
samples.insert(2, "sample_C")

print("Result:", samples)
```

Exercise 3: Given this list of voltage values
    voltages = [1.2, 0.8, 3.4, 2.1, 4.5, 0.5]
Solve the following tasks:

```python
# Task 1: Find and print the maximum and minimum voltages using built in functions
voltages = [1.2, 0.8, 3.4, 2.1, 4.5, 0.5]
print(max(voltages))
print(min(voltages))

# Task 2: Write a line of code that evaluates to true if 3.4 is present in voltages using the in operator
print(3.4 in voltages)

# Task 3: Calculate and print the average voltage using sum() and len()
average = sum(voltages) / len(voltages)
print(average)
```
